```python
from statsmodels.stats.proportion import proportions_ztest

# Using your aggregate results
normal_acc = 0.5229  # 52.29%
squeeze_acc = 0.4864  # 48.64%
n_paths = 10000

# Calculate number of correct predictions
normal_correct = int(normal_acc * n_paths)
squeeze_correct = int(squeeze_acc * n_paths)

# Proportion z-test
count = [normal_correct, squeeze_correct]
nobs = [n_paths, n_paths]

z_stat, p_value = proportions_ztest(count, nobs)

print("="*60)
print("PROPORTION Z-TEST (APPROXIMATE)")
print("="*60)
print(f"Normal accuracy:  {normal_acc:.2%} ({normal_correct}/{n_paths})")
print(f"Squeeze accuracy: {squeeze_acc:.2%} ({squeeze_correct}/{n_paths})")
print(f"Difference:       {(normal_acc - squeeze_acc)*100:.2f} percentage points")
print(f"\nZ-statistic:      {z_stat:.4f}")
print(f"P-value:          {p_value:.6f}")

if p_value < 0.05:
    print("\n✅ Degradation is STATISTICALLY SIGNIFICANT (p < 0.05)")
    print("   → The Non-linear Squeeze caused real model failure")
else:
    print("\n⚠️ Degradation not statistically significant")
```
